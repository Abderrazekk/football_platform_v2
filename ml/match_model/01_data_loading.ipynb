# NOTEBOOK : 01_data_loading.ipynb
# ROLE     : Read all raw Premier League match CSVs (2020-21 to 2026-27), add season column, and save one combined file.
# INPUT    : ../../data/raw/<season>/E0_Season_XXXXYYYY.csv
# OUTPUT   : ../../data/processed/all_matches_raw.csv

```python
from pathlib import Path
import pandas as pd

# Constants
RANDOM_STATE = 42
RAW_DATA_DIR = Path("../../data/raw")
PROCESSED_DATA_DIR = Path("../../data/processed")
OUTPUT_FILE = PROCESSED_DATA_DIR / "all_matches_raw.csv"

SEASONS = [
    "2020-21",
    "2021-22",
    "2022-23",
    "2023-24",
    "2024-25",
    "2025-26",
    "2026-27",
]

PROCESSED_DATA_DIR.mkdir(parents=True, exist_ok=True)
print("Paths configured successfully.")
```

```python
def load_season_matches(season_name):
    """Find and load the E0_Season_XXXXYYYY.csv file for one season."""
    season_dir = RAW_DATA_DIR / season_name
    csv_files = list(season_dir.glob("*Season_*.csv"))

    if len(csv_files) == 0:
        print(f"[WARNING] No match CSV found in {season_dir}")
        return None

    df = pd.read_csv(csv_files[0])
    df["season"] = season_name
    print(f"Loaded {season_name}: {len(df)} rows, {df.shape[1]} columns ({csv_files[0].name})")
    return df


all_dfs = []
for season in SEASONS:
    season_df = load_season_matches(season)
    if season_df is not None:
        all_dfs.append(season_df)

all_matches = pd.concat(all_dfs, ignore_index=True)
print(f"\nTotal combined matches: {len(all_matches)} rows, {all_matches.shape[1]} columns")
```

```python
all_matches.to_csv(OUTPUT_FILE, index=False)
print(f"Saved combined dataset to: {OUTPUT_FILE}")
all_matches.head()
```
