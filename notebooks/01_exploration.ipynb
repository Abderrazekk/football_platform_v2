```python
import sys
from pathlib import Path
sys.path.append(str(Path("../ml").resolve()))

import pandas as pd
import config

pd.set_option("display.max_columns", 200)
pd.set_option("display.width", 200)
print("Raw data folder:", config.DATA_RAW)
print("E0 files:", [p.name for p in sorted(config.DATA_RAW.rglob("E0_Season_*.csv"))])
```

```python
def find_file(filename):
    """Return the first matching file under data/raw, or None."""
    matches = list(config.DATA_RAW.rglob(filename))
    return matches[0] if matches else None


def inspect_file(path, label):
    """Print everything we need to know about one file."""
    print(f"\n{'='*70}\n{label}\nPath: {path}\n{'='*70}")
    if path is None:
        print("  >> FILE NOT FOUND")
        return None
    df = pd.read_csv(path)
    print(f"Shape: {df.shape}")
    print(f"\nColumns ({len(df.columns)}):\n{list(df.columns)}")
    print(f"\nDtypes:\n{df.dtypes}")
    print(f"\nMissing values:\n{df.isna().sum().sort_values(ascending=False)}")
    print(f"\nHead (2 rows):")
    display(df.head(2))
    print(f"\nDescribe (numeric):")
    display(df.describe())
    return df
```

```python
season = "2020-21"

e0            = inspect_file(find_file("E0_Season_20202021.csv"), "E0 matches")
gameweeks     = inspect_file(find_file("master_gameweeks.csv"),   "master_gameweeks")
player_gw     = inspect_file(find_file("master_player_gw.csv"),   "master_player_gw")
player_hist   = inspect_file(find_file("master_player_history.csv"), "master_player_history")
understat     = inspect_file(find_file("master_understat_teams.csv"), "master_understat_teams")
xP            = inspect_file(find_file("master_xP_projections.csv"),  "master_xP_projections")
```

```python
e0_files = sorted(config.DATA_RAW.rglob("E0_Season_*.csv"))
e0_shapes = {}
for path in e0_files:
    df = pd.read_csv(path)
    e0_shapes[path.stem] = {"rows": len(df), "cols": len(df.columns)}

print("Rows & columns per season:")
for name, info in e0_shapes.items():
    print(f"  {name:30s}  rows={info['rows']:>4}  cols={info['cols']}")

# Columns that differ between seasons
col_sets = [set(pd.read_csv(p, nrows=0).columns) for p in e0_files]
common = set.intersection(*col_sets)
all_cols = set.union(*col_sets)
print(f"\nColumns present in ALL seasons: {len(common)}")
print(f"Columns present in SOME seasons only: {len(all_cols - common)}")
print(f"  -> {sorted(all_cols - common)[:20]}")
```

```python
teams_e0 = set()
for path in config.DATA_RAW.rglob("E0_Season_*.csv"):
    df = pd.read_csv(path)
    teams_e0.update(df["Home_Team"].dropna().unique())
    teams_e0.update(df["Away_Team"].dropna().unique())

print(f"Teams found in E0 files ({len(teams_e0)}):")
for t in sorted(teams_e0):
    print(f"  {t}")

# FPL files use short names via 'team' column in master_gameweeks
if gameweeks is not None and "team" in gameweeks.columns:
    teams_fpl = sorted(gameweeks["team"].dropna().unique())
    print(f"\nTeams found in master_gameweeks ({len(teams_fpl)}):")
    for t in teams_fpl:
        print(f"  {t}")

if understat is not None and "team_name" in understat.columns:
    teams_us = sorted(understat["team_name"].dropna().unique())
    print(f"\nTeams found in master_understat_teams ({len(teams_us)}):")
    for t in teams_us:
        print(f"  {t}")
```

```python
# master_player_gw has 'element' (season-specific) + 'player_folder_name'
# master_player_history has 'element_code' (persistent across seasons)
# => we need a mapping element <-> element_code

# 1) Do two seasons of master_player_gw use the same 'element' for the same player?
gw_files = sorted(config.DATA_RAW.rglob("master_player_gw.csv"))
print(f"Found {len(gw_files)} master_player_gw.csv files")
for path in gw_files:
    df = pd.read_csv(path, usecols=["element", "player_folder_name"])
    print(f"  {path.parent.name if path.parent.name != 'raw' else 'root':>10}: "
          f"{df['element'].nunique()} unique elements, "
          f"{df['player_folder_name'].nunique()} unique player_folder_names")

# 2) Does master_player_history have element_code?
if player_hist is not None:
    print(f"\nplayer_history has element_code: {'element_code' in player_hist.columns}")
    print(f"Seasons covered in player_history: {sorted(player_hist['season_name'].unique())}")
```

```python
# Q1: is date-of-birth / age anywhere?
all_frames = [f for f in [e0, gameweeks, player_gw, player_hist, understat, xP] if f is not None]
birth_related = set()
for df in all_frames:
    for col in df.columns:
        if any(k in col.lower() for k in ["birth", "dob", "age"]):
            birth_related.add(col)
print(f"Q1. Birth/age columns found: {birth_related or 'NONE'}")

# Q2: E0 odds?
print(f"Q2. E0 has odds columns: {any('Odds' in c for c in e0.columns)}  "
      f"(e.g. {[c for c in e0.columns if 'Odds' in c][:3]})")

# Q3: player_gw contents?
needed = ["opponent_team", "was_home", "minutes", "total_points", "bps"]
present = [c for c in needed if c in player_gw.columns]
print(f"Q3. player_gw contains {len(present)}/{len(needed)} of {needed}: {present}")
# xG / xA?
xg_cols = [c for c in player_gw.columns if "xg" in c.lower() or "xa" in c.lower()]
print(f"    player_gw has xG/xA columns: {xg_cols or 'NONE'}")

# Q4: player_history = one row per player-season?
if player_hist is not None:
    dupes = player_hist.duplicated(subset=["player_folder_name", "season_name"]).sum()
    print(f"Q4. player_history duplicates (player, season): {dupes}  "
          f"(0 = one row per player per season ✓)")

# Q5: positions
if gameweeks is not None and "position" in gameweeks.columns:
    print(f"Q5. Positions in master_gameweeks: {sorted(gameweeks['position'].unique())}")
```
